# M3: Running library implementations of other machine learning models

```python
import os
import numpy as np
from sklearn.neighbors import KNeighborsClassifier as KNN
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier as RFC
from sklearn.linear_model import LogisticRegression as LR
from sklearn.neural_network import MLPClassifier as MLPC

import tensorflow as tf
from tensorflow.keras import models, layers
from keras.optimizers import SGD
```

```python
# Load post-PCA subsets
data_folder = 'data/post_pca_subsets'
PCA_full_X_train = np.load(os.path.join(data_folder, 'PCA_full_X_train.npy'))
PCA_full_y_train = np.load(os.path.join(data_folder, 'PCA_full_y_train.npy'))
PCA_X_test = np.load(os.path.join(data_folder, 'PCA_X_test.npy'))
PCA_y_test= np.load(os.path.join(data_folder, 'PCA_y_test.npy'))
```

# K Neigbours Classifier

```python
#Initialise the machine
KNeNe = KNN(weights = 'distance', n_neighbors= 5, p=1 )
KNeNe.fit(PCA_full_X_train, PCA_full_y_train)

#Get test data predictions
KNNtest_predictions = KNeNe.predict(PCA_X_test)

#Get percentage of correct predictions
KNNcorrect_pred_count = sum(KNNtest_predictions == PCA_y_test)
KNNratio_of_correct = KNNcorrect_pred_count/len(PCA_y_test)
print(f'Percentage of correct categorisations is', KNNratio_of_correct *100, '%')
```

# Random Forest Classifier

```python
#Initialise the machine
Rando = RFC(n_estimators= 100, criterion = 'entropy', max_depth = 100)
Rando.fit(PCA_full_X_train, PCA_full_y_train)

#Get validation data predictions
RFCtest_predictions = Rando.predict(PCA_X_test)

#Get percentage of correct predictions
RFCcorrect_pred_count = sum(RFCtest_predictions == PCA_y_test)
RFCratio_of_correct = RFCcorrect_pred_count/len(PCA_y_test)
print(f'Percentage of correct categorisations is', RFCratio_of_correct *100, '%')
```

# Support Vector Classification

```python
#Initialise the machine
SupVecMac = SVC(C=10, kernel = 'rbf')
SupVecMac.fit(PCA_full_X_train, PCA_full_y_train)

#Get validation data predictions
SVCtest_predictions = SupVecMac.predict(PCA_X_test)

#Get percentage of correct predictions
SVCcorrect_pred_count = sum(SVCtest_predictions == PCA_y_test)
SVCratio_of_correct = SVCcorrect_pred_count/len(PCA_y_test)
print(f'Percentage of correct categorisations is', SVCratio_of_correct *100, '%')
```

# Logistic Regression

```python
#Initialise the machine
LogReg = LR(C=1, multi_class = 'ovr', penalty = 'l2')
LogReg.fit(PCA_full_X_train, PCA_full_y_train)

#Get validation data predictions
LRtest_predictions = LogReg.predict(PCA_X_test)

#Get percentage of correct predictions
LRcorrect_pred_count = sum(LRtest_predictions == PCA_y_test)
LRratio_of_correct = LRcorrect_pred_count/len(PCA_y_test)
print(f'Percentage of correct categorisations is', LRratio_of_correct *100, '%')
```

# Multi-Layer Perceptron (Pretty sure this is the dame as a Feed Forward Neural Network)

```python
#Initialise the machine
MLP = MLPC(activation = 'relu', hidden_layer_sizes = [100])
MLP.fit(PCA_full_X_train, PCA_full_y_train)

#Get validation data predictions
MLPtest_predictions = MLP.predict(PCA_X_test)

#Get percentage of correct predictions
MLPcorrect_pred_count = sum(MLPtest_predictions == PCA_y_test)
MLPratio_of_correct = MLPcorrect_pred_count/len(PCA_y_test)
print(f'Percentage of correct categorisations is', MLPratio_of_correct *100, '%')
```
